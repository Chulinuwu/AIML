```python
from pyswip import Prolog

prolog = Prolog()

prolog.assertz("man(marcus)")
prolog.assertz("pompeian(marcus)")
prolog.assertz("roman(X) :- pompeian(X)")
prolog.assertz("ruler(caesar)")
prolog.assertz("loyalto(X, caesar) :- roman(X)")
prolog.assertz("hate(X, caesar) :- roman(X)")
prolog.assertz("loyalto(X, f1(X))")
prolog.assertz("hate(X, Y) :- man(X), ruler(Y), tryassassinate(X, Y), \\+ loyalto(X, Y)")
prolog.assertz("tryassassinate(marcus, caesar)")

result = list(prolog.query("hate(marcus, caesar)"))

# แสดงผลลัพธ์
if result:
    print("Marcus hate Caesar: TRUE")
else:
    print("Marcus hate Caesar: FALSE")
```
